```python
## 16. Loading the Saved Model
import pickle
import pandas as pd

filename = "LightGBM_Regression_Model.sav"
load_model = pickle.load(open(filename, 'rb'))
print("Model loaded successfully.")
```

```python
## 17. Taking New Customer Input
age = float(input("Age: "))
bmi = float(input("BMI: "))
children = float(input("Children: "))
sex_male = int(input("Sex Male (0 or 1): "))
smoker_yes = int(input("Smoker Yes (0 or 1): "))
new_customer = [[age, bmi, children, sex_male, smoker_yes]]
```

```python
## 18. Generating the Insurance Charge Prediction
prediction = load_model.predict(new_customer)
print("Predicted Insurance Charges:", prediction[0])
```
